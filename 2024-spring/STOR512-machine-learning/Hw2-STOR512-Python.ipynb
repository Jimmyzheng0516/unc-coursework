```python
import numpy as np

# Calculate Euclidean distance between two points
def euclidean_dist(pt1, pt2):
    return np.sqrt(np.sum((pt1 - pt2) ** 2)) # For any two d*1 vectors, i.e., point 1 and point 2

# Function to calculate and find the maximum distance
def calc_dist(X):
    n = X.shape[1] # Set n = number of points
    max_dist = 0
    pt1_max = 1 # Initialize two points with the largest distance
    pt2_max = 2
    
    for i in range(n-1): 
        for j in range(i+1, n):
            dist = euclidean_dist(X[:,i], X[:,j])
            if dist > max_dist:
                max_dist = dist
                pt1_max = i + 1 
                pt2_max = j + 1 # Report two points with the largest distance
                
    return {"max_dist": max_dist, "pt1": pt1_max, "pt2": pt2_max}

# Run the function with the given data
X = np.array([[4, 1, 5],
              [3, 2, -4],
              [-5, 0, -6],
              [7, 9, -3],
              [9, 10, -8]]).T

result = calc_dist(X)

print(result)
```

```python
import numpy as np

def power_method(A, max_iter=20):
    b = np.array([1, 1])  # Initialize the seed vector b(0)
    
    for i in range(max_iter):
        b = np.dot(A, b)
        b = b / np.sqrt(np.sum(b**2))  # Normalize b
    
    ATA = np.dot(A.T, A)
    eigen_max = np.dot(np.dot(b.T, ATA), b)  # Compute the maximum eigenvalue (approximated)
    
    return np.sqrt(eigen_max)  # Compute the spectral norm

# Apply the function to the matrix A
A = np.array([[1, 2], [2, -1]])

print(power_method(A))
```

```python
import numpy as np

def g_s(A):
  
    d, n = A.shape
    Q = np.zeros((d, n))
    
    for i in range(n):
      
        v_i = A[:, i]
        u_i = v_i.copy()
        
        for j in range(i):
          
            u_j = Q[:, j]
            proj = np.dot(v_i, u_j) / np.dot(u_j, u_j) * u_j
            u_i -= proj
            
        Q[:, i] = u_i / np.linalg.norm(u_i)
        
    return Q

# Apply the Gram-Schmidt process to matrix A
A = np.array([
    [1.5, 3, 0, 1],
    [2, 1.5, 0, 2],
    [1, 5.5, 1, 1],
    [0, 2, 2, 3]
])

g_s(A)
```

```python
import numpy as np

# Given vectors
p = np.array([0.2, 0.3, 0.1, 0.4])
x = np.array([1, 2, -1, 4])

e_x = np.exp(x)

sum_e_x = np.sum(e_x)

# Calculate the gradient
grad = p - (e_x / sum_e_x)

# Calculate the Hessian
d = len(x)  
H = np.zeros((d, d))  # Initialize the matrix

for i in range(d):
    for j in range(d):
        if i == j:
            H[i, j] = (e_x[i] / sum_e_x) * ((e_x[i] / sum_e_x)-1)
        else:
            H[i, j] = (e_x[i] * e_x[j]) / (sum_e_x**2)

grad, H
```
